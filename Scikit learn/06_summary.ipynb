```python
# function transformet to count no of words

def count_words(reviews):
    return np.array([len(review.split()) for review in reviews]).reshape(-1, 1)
```

```python
from sklearn.preprocessing import FunctionTransformer

word_count_transformer = FunctionTransformer(count_words)
```

```python
# feature union -> bag of words and word count to be implemented on review col

from sklearn.pipeline import FeatureUnion
from sklearn.feature_extraction.text import CountVectorizer

feature_union = FeatureUnion([
    ('bow' , CountVectorizer()),
    ('word_count' , word_count_transformer)
])
```

```python
# col transformer

from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

column_transform = ColumnTransformer(transformers=
                                    [
                                        ('platform_ohe' , OneHotEncoder() , ['Social Media Platform']),
                                        ('review_fu' , feature_union , ['Review']),
                                        ('age_impu' , SimpleImputer() , ['age'])
                                    ], remainder= 'drop')
```

```python
# now implement the final pipeline and at last obtain a trained model via logistic regression
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import MaxAbsScaler
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.pipeline import Pipeline

final_pipeline = Pipeline(steps=
                           [
                               ('col_transformer', column_transform),
                               ('scaler', MaxAbsScaler()),
                               ('selector', SelectKBest(score_func=chi2,k=10)),
                               ('classifier', LogisticRegression())
                           ]
)
```
