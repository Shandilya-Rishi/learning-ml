```python
# to create an empty data frame
df = pd.DataFrame()
```

```python
# to extract the complete dataset from the api need to perfom the action in loop 
# and keep appending the data into the data frame
for i in range(1,536):
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=d93b271401f5d2e4bf17b81e482475d6&page={}'.format(i))
    temp_df = pd.DataFrame(response.json()[['id' , 'title' , 'overview']])
    df = df.append(temp_df , ignore_index = True) # ignore_index : so that repeared indices do not get appended in the df
```
