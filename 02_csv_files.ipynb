## CSV file from URL


```python
import requests 
from io import StringIO

url = "https://people.sc.fsu.edu/~jburkardt/data/csv/addresses.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url , headers=headers) # all data from usrl stores in here

df2 = pd.read_csv(StringIO(req.text))
print(df2)
```

## Converters


```python
def rename(name):
    if name == "Royal Challengers Banglore":
        return 'RCB'
    else:
        return name
```

```python
rename("Royal Challengers Banglore")
```

```python
pd.read_csv("IPL Matches 22 .csv" , converters = {'team1' : rename}) 
```

## na_values : letting the compuler know the values to be treated like na (not available) values e.g. : - hyphen

```python
pd.read_csv("pokemon.csv" , na_values= 'Poison')
```
